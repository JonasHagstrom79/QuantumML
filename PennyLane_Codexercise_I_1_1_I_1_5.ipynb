```python
import numpy as np

ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

def normalize_state(alpha, beta):
    """Compute a normalized quantum state given arbitrary amplitudes."""

    # 1. Skapa vektorn genom att multiplicera alpha/beta med basvektorerna
    # Detta ger vektorn [alpha, beta]
    vector = alpha * ket_0 + beta * ket_1

    # 2. Beräkna längden (normen) med rätt funktion
    length = np.linalg.norm(vector)

    # 3. Dela vektorn med längden för att normalisera den
    v_norm = vector / length

    return v_norm
```
