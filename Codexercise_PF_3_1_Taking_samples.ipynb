```python
dev = qml.device("default.qubit", wires = 2, shots = 1) # Wires = numbers of qbits


@qml.qnode(dev)
def circuit():
    """
    This quantum function implements the circuit shown above
    and should return a sample from all qubits
    """

    qml.H(wires=0)
    qml.CNOT(wires=[0,1])

    return qml.sample()

print(circuit())
```

```python
qml.draw_mpl(circuit)()
plt.show()
```
