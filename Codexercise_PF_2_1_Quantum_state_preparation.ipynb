I PennyLane (och kvantfysik generellt) kan du inte skriva in symbolen |001> direkt i koden. Istället låter man positionen i en lista representera de olika tillstånden.

3 qubits, finns det totalt 23=8 möjliga tillstånd (från 000 till 111). För att sätta dina värden (alpha, beta, gamma) på rätt plats, måste vi bygga en lista med 8 platser och lägga värdena på exakt rätt index.

Kvanttillstånden räknas binärt (som vanliga datorsiffror):

    Plats 0: |000> → Vi vill inte ha denna, så värdet blir 0.

    Plats 1: |001> → Här ska alpha vara

    Plats 2: |010> → Här ska beta vara

    Plats 3: |011> → 0

    Plats 4: |100> → Här ska gamma vara

    Plats 5: |101> → 0

    Plats 6: |110> → 0

    Plats 7: |111> → 0

```python
dev = qml.device("default.qubit", wires = 3)

@qml.qnode(dev)
def prep_circuit(alpha, beta, gamma):
    """
    Prepares the state alpha|001> + beta|010> + gamma|100>.
    Args:
    alpha, beta, gamma (np.complex): The coefficients of the quantum state
    to prepare.
    Returns:
    (np.array): The quantum state
    """
    state_vector = [0, alpha, beta, 0, gamma, 0, 0, 0]
    qml.StatePrep(state_vector, wires=range(3), normalize=True)


    return qml.state()

alpha, beta, gamma = 1/np.sqrt(3), 1/np.sqrt(3), 1/np.sqrt(3),

print("The prepared state is", prep_circuit(alpha, beta, gamma))
```
